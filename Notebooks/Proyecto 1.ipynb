## 2. Entendimiento del problema

**Pregunta analítica:**

¿Cómo se relaciona la edad con el nivel de satisfacción con la vida de las personas de 18 años o más que vivían en Bogotá hace 12 meses?

**Información necesaria para responder la pregunta:**

Para responder esta pregunta, antes de consultar la base de datos se considera necesario conocer la edad de las personas, con el fin de identificar a la población adulta y posteriormente comparar diferentes grupos de edad. También se necesita conocer el nivel de satisfacción con la vida, ya que esta será la medida utilizada para analizar el bienestar subjetivo de la población.
Además, se requiere información que permita identificar dónde vivía cada persona hace 12 meses, para determinar cuáles corresponden a la población de Bogotá. De esta manera, será posible observar y comparar los niveles de satisfacción con la vida entre personas de diferentes edades dentro de la población adulta definida para el análisis.


## 3. Preparación de los datos (ETL)


```python
pd.set_option('display.max_columns', None)

df = pd.read_csv("../Data/Raw/Características y composición del hogar.csv", sep=";", encoding="utf-8")

variables_interes = ['DIRECTORIO', 'P6040', 'P1895', 'P6074', 'P753', 'P753S1', 'P753S2']
df_sel = df[variables_interes].copy()
print("Dimensiones df_sel:", df_sel.shape)
df_sel.head()
```

### 3.2 Filtro: personas de 18+ años residentes en Bogotá (últimos 12 meses)


```python
df_adultos = df_sel[df_sel['P6040'] >= 18].copy()
print("Personas de 18+ años (a nivel nacional, antes de filtrar por Bogotá):", df_adultos.shape[0])
print("Filtro de residencia pendiente de aplicar (falta cruzar con 'Datos de la vivienda').")
```

### 3.3 Calidad de los datos


```python
resumen_calidad = pd.DataFrame({
    'nulos': df_adultos[['P6040', 'P1895', 'P6074', 'P753']].isna().sum(),
    '% nulos': (df_adultos[['P6040', 'P1895', 'P6074', 'P753']].isna().mean() * 100).round(2),
})
resumen_calidad
```

```python
print("Distribución de P1895 (satisfacción con la vida) en adultos 18+:")
print(df_adultos['P1895'].value_counts(dropna=False).sort_index())
```

```python
print("Distribución de P6074 (¿siempre ha vivido en este municipio?) en adultos 18+:")
print(df_adultos['P6074'].value_counts(dropna=False))
print()
print("De quienes se mudaron (P6074 = 2), a dónde se mudaron (P753):")
print(df_adultos.loc[df_adultos['P6074'] == 2, 'P753'].value_counts(dropna=False))
```

### 4.2 Variables categóricas
Frecuencia de las categorías.

```python
bins = [18, 29, 44, 59, 120]
labels = ['18-29', '30-44', '45-59', '60+']
df_adultos['grupo_edad'] = pd.cut(df_adultos['P6040'], bins=bins, labels=labels, right=True, include_lowest=True)

df_adultos['grupo_edad'].value_counts(normalize=True).sort_index().round(3) * 100
```

```python
df_adultos.groupby('grupo_edad', observed=True)['P1895'].agg(['count', 'mean', 'median']).round(2)
```

### 4.4 Valores atípicos (outliers)


```python
Q1 = df_adultos['P6040'].quantile(0.25)
Q3 = df_adultos['P6040'].quantile(0.75)
IQR = Q3 - Q1
limite_inf = Q1 - 1.5 * IQR
limite_sup = Q3 + 1.5 * IQR

outliers_edad = df_adultos[(df_adultos['P6040'] < limite_inf) | (df_adultos['P6040'] > limite_sup)]
print(f"Límite superior (IQR): {limite_sup}")
print("Cantidad de outliers en edad:", outliers_edad.shape[0])
outliers_edad[['P6040', 'P1895']]
```

## 5. Hallazgos

> ¿Qué le pueden decir a la Alcaldía sobre el tema analizado? ¿Cómo responden estos hallazgos a la
> pregunta analítica? 

A partir del análisis descriptivo se identificaron varios resultados que permiten comprender mejor la relación entre la edad y la satisfacción con la vida en la población adulta analizada.

A continuación se presentan los principales hallazgos junto con los cálculos y visualizaciones que los respaldan.

### **Hallazgo 1** / La satisfacción con la vida es generalmente alta

Primero se revisó el nivel general de satisfacción con la vida de las personas de 18 años o más. Para esto se calcularon la media, la mediana y el porcentaje de personas que reportaron valores entre 7 y 10 en la escala de satisfacción.

```python
media_satisfaccion = df_adultos['P1895'].mean()
mediana_satisfaccion = df_adultos['P1895'].median()
porcentaje_7_10 = (df_adultos['P1895'] >= 7).mean() * 100

print("Media de satisfacción:", round(media_satisfaccion, 2))
print("Mediana de satisfacción:", round(mediana_satisfaccion, 2))
print("Porcentaje con satisfacción entre 7 y 10:", round(porcentaje_7_10, 2), "%")
```

```python
frecuencia_satisfaccion = df_adultos['P1895'].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(7, 4))

frecuencia_satisfaccion.plot(kind='bar', ax=ax)

ax.set_title("Distribución de la satisfacción con la vida")
ax.set_xlabel("Satisfacción con la vida (0-10)")
ax.set_ylabel("Número de personas")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

La población adulta analizada presenta niveles generalmente altos de satisfacción con la vida. El promedio es de 8,14 sobre 10, la mediana es 8 y aproximadamente el 86,64 % de las personas reportó una satisfacción entre 7 y 10.

La gráfica también muestra que las respuestas se concentran principalmente en la parte alta de la escala, especialmente en los valores 8, 9 y 10.

### **Hallazgo 2** / La satisfacción disminuye ligeramente en los grupos de mayor edad

Para observar con mayor claridad la relación entre edad y satisfacción, se dividió la población en cuatro grupos de edad y se comparó el promedio y la mediana de satisfacción con la vida en cada uno.

```python
resumen_grupos = df_adultos.groupby('grupo_edad', observed=True)['P1895'].agg(['count', 'mean', 'median']).round(2)

resumen_grupos
```

```python
promedio_grupos = df_adultos.groupby('grupo_edad', observed=True)['P1895'].mean()

diferencia_extremos = promedio_grupos.iloc[0] - promedio_grupos.iloc[-1]

print("Diferencia entre 18-29 y 60+:", round(diferencia_extremos, 2))
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

promedio_grupos.plot(kind='bar', ax=ax)

ax.set_title("Satisfacción promedio con la vida por grupo de edad")
ax.set_xlabel("Grupo de edad")
ax.set_ylabel("Satisfacción promedio (0-10)")
ax.set_ylim(0, 10)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Se observa una disminución gradual de la satisfacción promedio a medida que aumenta la edad. El promedio es de 8,24 en las personas de 18 a 29 años, 8,20 entre 30 y 44 años, 8,15 entre 45 y 59 años y 7,99 en las personas de 60 años o más.

Sin embargo, la diferencia entre el grupo más joven y el de mayor edad es de aproximadamente 0,25 puntos en una escala de 0 a 10. Por lo tanto, aunque existe una tendencia descendente, la diferencia entre los grupos es pequeña.

### **Hallazgo 3** / La proporción de personas con satisfacción entre 7 y 10 también disminuye con la edad

Para complementar la comparación de promedios, se calculó qué porcentaje de personas de cada grupo de edad reportó una satisfacción entre 7 y 10.

```python
df_adultos['satisfaccion_7_10'] = df_adultos['P1895'] >= 7

porcentaje_grupos = (df_adultos.groupby('grupo_edad', observed=True)['satisfaccion_7_10'].mean() * 100).round(2)

porcentaje_grupos
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

porcentaje_grupos.plot(kind='bar', ax=ax)

ax.set_title("Personas con satisfacción entre 7 y 10 por grupo de edad")
ax.set_xlabel("Grupo de edad")
ax.set_ylabel("Porcentaje de personas")
ax.set_ylim(0, 100)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

El porcentaje de personas que reporta una satisfacción entre 7 y 10 también disminuye gradualmente con la edad; este porcentaje pasa de 88,15 % en el grupo de 18 a 29 años a 83,93 % en las personas de 60 años o más.

Los grupos de 30 a 44 y de 45 a 59 años presentan porcentajes intermedios de 87,59 % y 86,95 %, respectivamente. Este resultado sigue el mismo patrón observado al comparar los promedios de satisfacción.

### **Hallazgo 4** / La relación entre edad y satisfacción es muy débil

Aunque los resultados anteriores muestran una pequeña disminución de la satisfacción en los grupos de mayor edad, es importante revisar qué tan fuerte es la relación entre ambas variables.

```python
correlacion_edad_satisfaccion = df_adultos['P6040'].corr(df_adultos['P1895'])

print("Correlación entre edad y satisfacción:",round(correlacion_edad_satisfaccion, 3))
```

La correlación entre edad y satisfacción con la vida es aproximadamente -0,064 ; el signo negativo coincide con la disminución observada en los grupos de edad, sin embargo, el valor se encuentra muy cerca de cero, por lo que podemos ver que la relación entre ambas variables es muy débil.

### **Respuesta a la pregunta analítica**

Los resultados muestran que la satisfacción con la vida disminuye ligeramente a medida que aumenta Los resultados muestran que la satisfacción con la vida baja un poco a medida que aumenta la edad. Sin embargo, las diferencias entre los grupos son pequeñas y la relación entre edad y satisfacción es muy débil, además, esta relación no significa que la edad cause una menor satisfacción con la vida, ya que correlación no implica causalidad.

**¿Qué no se puede concluir?**  - Estos resultados no permiten afirmar que una mayor edad cause una menor satisfacción con la vida, ya que se trata de una asociación y pueden intervenir otros factores; además tampoco puede afirmarse que la edad sea suficiente para explicar las diferencias en bienestar entre las personas. Asimismo, al aplicarse únicamente el filtro de edad y no el de lugar de residencia los resultados describen a la población adulta de la base analizada y no pueden interpretarse como resultados exclusivos de la población de Bogotá. 

**¿Qué haría falta para profundizar?**  - Sería útil incorporar otros elementos de bienestar disponibles en la ECV, como salud, ingreso, seguridad, trabajo, etc. Analizar como estas nuevas dimensiones se relacionan juntamente con la edad y la satisfacción con la vida; a su vez, necesitaríamos contar con una muestra más amplia específicamente de residentes de Bogotá si se quisiera realizar conclusiones de esta ciudad.
